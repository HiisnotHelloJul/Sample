```python
# Distance Calculator


# Find a predetermined library to easily code the formula
import math

# State the variables clearly
# x1 and x2 are the domains and y1 and y2 are the ranges

# Ask the user to enter the coordinates of the two points
x1 = int(input ("x1: "))
x2 = int(input ("x2: "))
y1 = int(input ("y1: "))
y2 = int(input ("y2: "))

# Finding the Formula
# d = square root of ((x2 - x1)^2 + (y2 - y1)^2)

# Using the formula as a code
distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))

# Display the distance and round it to two decimal points
print("The distance between the two points is:", round(distance, 2))
```
